```python
from typing import List, Tuple, Optional

class Node:
    def __init__(self, name: str, children: Optional[List['Node']] = None, value: Optional[int] = None):
        self.name = name
        self.children = children if children else []
        self.value = value

def alpha_beta(node: Node, alpha: float, beta: float, maximizing_player: bool,
               path: List[str], pruned_paths: List[List[str]]) -> Tuple[float, List[str]]:
    """
    Performs Alpha-Beta pruning on a game tree.
    Returns a tuple of (best_value, best_path_to_leaf).
    """
    if node.value is not None:
        return node.value, path + [node.name]

    current_path = path + [node.name]

    if maximizing_player:
        max_eval = float('-inf')
        best_path = []
        for i, child in enumerate(node.children):
            eval_value, eval_path = alpha_beta(child, alpha, beta, False, current_path, pruned_paths)
            if eval_value > max_eval:
                max_eval = eval_value
                best_path = eval_path
            alpha = max(alpha, eval_value)

            if beta <= alpha:
                for remaining_child in node.children[i + 1:]:
                    pruned_paths.append(current_path + [remaining_child.name])
                break
        return max_eval, best_path
    else:
        min_eval = float('inf')
        best_path = []
        for i, child in enumerate(node.children):
            eval_value, eval_path = alpha_beta(child, alpha, beta, True, current_path, pruned_paths)
            if eval_value < min_eval:
                min_eval = eval_value
                best_path = eval_path
            beta = min(beta, eval_value)

            if beta <= alpha:
                for remaining_child in node.children[i + 1:]:
                    pruned_paths.append(current_path + [remaining_child.name])
                break
        return min_eval, best_path

def build_tree_automatically() -> Node:
    """
    Interactively builds a tree by prompting ONLY for branching factor and leaf values.
    Names nodes automatically.
    """
    node_counter = 1

    def build_node(parent_name: str) -> Node:
        nonlocal node_counter
        is_leaf_input = input(f"Is the current child of {parent_name} a leaf node? (y/n): ").strip().lower()
        if is_leaf_input == 'y':
            val = int(input("  Enter integer value for this leaf: "))
            leaf_name = f"Leaf_{node_counter}"
            node_counter += 1
            return Node(leaf_name, value=val)
        else:
            num_children = int(input(f"  Enter number of children: "))
            internal_name = f"Node_{node_counter}"
            node_counter += 1
            children = []
            for i in range(num_children):
                children.append(build_node(internal_name))
            return Node(internal_name, children=children)

    return build_node("Start")

# Build the tree dynamically using user input (values only)
tree_root = build_tree_automatically()

pruned_branches = []
optimal_value, optimal_path = alpha_beta(tree_root, alpha=float('-inf'), beta=float('inf'),
                                         maximizing_player=True, path=[], pruned_paths=pruned_branches)

print("\n--- Execution Results ---")
print(f"Optimal value at Root: {optimal_value}")
print(f"Optimal Path: {' -> '.join(optimal_path)}")
print("\nPruned Branches:")
if pruned_branches:
    for p in pruned_branches:
        print(" -> ".join(p) + " -> ... (Pruned)")
else:
    print("No pruning occurred.")
```
